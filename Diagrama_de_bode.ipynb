```python
import numpy as np
import matplotlib.pyplot as plt

# Función de transferencia del RLC serie: Vc/Vin
def H(w, R, L, C):
    return 1 / (1 - L*C*w**2 + 1j*R*C*w)

# Vector de frecuencias (logarítmico)
w = np.logspace(-1, 5, 2000)

# Evaluar H(w)
Heval = H(w, R=1000, L=180e-3, C=120e-6)

# Magnitud y fase
Hmag = 20 * np.log10(np.abs(Heval))
Hphase = np.angle(Heval, deg=True)

# =======================
#       GRÁFICO BODE
# =======================

plt.figure(figsize=(10, 7))

# --- MAGNITUD ---
plt.subplot(2, 1, 1)
plt.semilogx(w, Hmag, linewidth=2)
plt.grid(which="both", linestyle="--", alpha=0.6)
plt.title("Diagrama de Bode - Circuito RLC Serie")
plt.ylabel("Magnitud (dB)")

# --- FASE ---
plt.subplot(2, 1, 2)
plt.semilogx(w, Hphase, linewidth=2)
plt.grid(which="both", linestyle="--", alpha=0.6)
plt.xlabel("Frecuencia (rad/s)")
plt.ylabel("Fase (°)")

plt.tight_layout()
plt.show()
```
